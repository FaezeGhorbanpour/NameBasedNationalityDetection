### Saving the data

```python
data['country'] = data.iso3.apply(lambda x: iso3_to_country(x))
```

###### Standardizing country names

```python
data.country = data.country.apply(lambda x: country_name_standardizer(x))
```

```python
data['full-name'] = data['name']
```

```python
data['id'] = data.apply(lambda x:'wikidata-'+str(x.name), axis=1)
```

```python
data.head()
```

```python
processed_data = data[['id', 'full-name', 'country']]
processed_data.set_index('id', inplace=True)
```

```python
wikidata_.save_data(processed_data)
```
