### Saving the data

```python
data = data[data.country != '']
data = data[data.name != '']
```

```python
data['id'] = data.apply(lambda x: 'kaggle-' + str(x.name), axis=1)
```

```python
data['full-name'] = data.name
```

```python
from Datasets.mapping import iso2_to_country

data['country'] = data.country.apply(lambda x: iso2_to_country(x))
```

```python
data = data[['id', 'full-name', 'country']]
data.set_index('id', inplace=True)
data.head()
```

```python
kaggle.save_data(data)
```
