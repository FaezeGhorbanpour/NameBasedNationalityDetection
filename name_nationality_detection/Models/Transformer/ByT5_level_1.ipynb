```python
from Datasets.datasets import Dataset
import os
os.environ['NationalityDatasets'] = "./data/"

dt = Dataset()
data = dt.get_data()
train = dt.get_train()[:10000]
test = dt.get_test()[:1000]
validation = dt.get_validation()[:1000]
train_validation = dt.get_train_validation()
```

```python
labels = list(train['level_1'].unique())
labels
```

```python
model_dir = "./data/"
```

```python
from torch.utils.data import DataLoader
import torch
from Models.Transformer.ByT5 import model_init, MyDataset, DataCollator, Pipeline
from transformers import AutoTokenizer

checkpoint = "google/byt5-small"
tokenizer = AutoTokenizer.from_pretrained(checkpoint)
model = model_init(checkpoint)
optimizer = torch.optim.AdamW(model.parameters(), lr=1e-4)
train_ds = MyDataset(train, 'level_1')
val_ds = MyDataset(validation, 'level_1')
data_collator = DataCollator(tokenizer)
train_loader = DataLoader(train_ds, batch_size=128, collate_fn=data_collator)
val_loader = DataLoader(val_ds, batch_size=128, collate_fn=data_collator)
```

```python
@torch.no_grad()
def eval_model(model, val_loader, tokenizer):
    model.eval()
    total = 0
    count = 0
    stream = tqdm(val_loader)
    for batch in stream:
        batch = dict_to_cuda(batch)
        loss = model(**batch).loss
        total += loss.item()
        count += 1
        stream.set_description_str(desc=f"val_edit_dist {total / count:.3f}")

    model.train()
    return total / count
def dict_to_cuda(batch):
    return {k: v.cuda() for k, v in batch.items()}
```

```python
from datasets import tqdm

max_epochs = 5
best = float("inf")
trigger = 0

for epoch in range(max_epochs):
    stream = tqdm(train_loader)
    total = 0
    count = 0
    for batch in stream:
        batch = dict_to_cuda(batch)
        loss = model(**batch).loss
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        total += loss.item()
        count += 1
        stream.set_description_str(
            f"Epoch {epoch}/{max_epochs} | Loss {total / count : .3f}"
        )
    loss = total / count
    eval_metric = eval_model(model, val_loader, tokenizer)

    print(
        f"Epoch {epoch}/{max_epochs} | Loss {loss: .3f} | Val Loss {eval_metric: .4f}"
    )

    trigger += 1

    if eval_metric < best:
        model.save_pretrained(model_dir)
        best = eval_metric
        trigger = 0

    if trigger > 5:
        break
```

```python
pipeline = Pipeline(model, tokenizer, device="cuda")
print(pipeline.generate("Faeze Ghorbanpour"))
print(pipeline.generate("Ali Mohammadi"))
print(pipeline.generate("Eugenio Paglino"))
print(pipeline.generate("Victor hugo"))
print(pipeline.generate("Goran Glavaš"))
```

```python
outputs = []
stream = tqdm(test["name"])
for text in stream:
    output = pipeline.generate(text)
    # stream.set_description_str(output)
    outputs.append(output)
```

```python

test["pred"] = outputs
test.to_csv("./data/", index=False)
```
