```python
from Datasets.datasets import Dataset
from Models.NeuralNetworks.char_cnn.char_cnn import CharCNNConfig, CharCNN
from Models.NeuralNetworks.data_loader import DatasetLoader
from Models.NeuralNetworks.neural_network import NeuralModel
import torch
import os
os.environ['NationalityDatasets'] = "./data/"
torch.cuda.empty_cache()
```

### Loading Config

```python
config = CharCNNConfig()
config.num_conv_layers = 6
config.batch_size = 1024 * 5
```

### Loading Dataset

```python
dt = Dataset()
```

```python
data = dt.get_data()
train = dt.get_train()
test = dt.get_test()
validation = dt.get_validation()
```

```python
dl = DatasetLoader(data, 'level_2', config)
```

```python
train_loader = dl.get_data_loader(train, True)
test_loader = dl.get_data_loader(test)
validation_loader = dl.get_data_loader(validation)
```

### Loading Model

```python
model = CharCNN
```

#### Learning


```python
nn = NeuralModel(name='char_cnn', column='level_2', model_type=model, config=config)
```

```python
nn.optuna_learner(train_loader, validation_loader)
```
