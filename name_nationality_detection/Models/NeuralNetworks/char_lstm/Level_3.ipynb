```python
from Datasets.datasets import Dataset
from Models.NeuralNetworks.char_lstm.char_lstm import CharLSTMConfig, CharLSTM
from Models.NeuralNetworks.data_loader import DatasetLoader
from Models.NeuralNetworks.neural_network import NeuralModel
import torch
import os
os.environ['NationalityDatasets'] = "./data/"
torch.cuda.empty_cache()
```

### Loading Config

```python
config = CharLSTMConfig()
```

### Loading Dataset

```python
dt = Dataset()
```

```python
data = dt.get_data()
train = dt.get_train()
test = dt.get_test()
validation = dt.get_validation()
```

```python
dl = DatasetLoader(data, 'level_3', config)
```

```python
train_loader = dl.get_data_loader(train, True)
test_loader = dl.get_data_loader(test)
validation_loader = dl.get_data_loader(validation)
```

### Loading Model

```python
model = CharLSTM
```

#### Learning


```python
nn = NeuralModel(name='char_lstm', column='level_3', model_type=model, config=config)
```

```python
%%timeit
nn.learner(train_loader, validation_loader)
```

```python
# nn.optuna_learner(train_loader, validation_loader)
```

```python
predicteds, y_trues = nn.testing(test_loader, )
```

```python
nn.classification_metrics(pred=predicteds, label=y_trues)
```

```python
cm = nn.confusion_matrix(label=y_trues, pred=predicteds)
nn.display_confusion_matrix(cm)
```

```python
nn.classification_report(label=y_trues, pred=predicteds)
```
